# This Notebook does all Exploratory Data Analysis needed to understand the data, frame the problem, and prepare the data before training

```python
# Importing Libaries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Importing Dataframe
df = pd.read_csv("c:/Users/DARA2/Documents/GitHub/modelops-lab/data/raw/WineQT.csv")
```

```python
# Understanding the dataframe

# Features in the data
print("Features in the dataset: \n", df.columns.tolist())

# Data Types
print("Data Types in the dataset:\t", df.dtypes)

# Missing Values and duplicates
print("Total Missing Values: ", df.isnull().sum())
print("Total Duplicated Values: ", df.duplicated().sum())


# Summary Statistics
print("Summary statistics: \n", df.describe())
```

```python
# Boxplots to check for outliers

for column in df.columns:
    if df[column].dtype != 'object':
        plt.figure(figsize=(8, 6))
        plt.boxplot(df[column])
        plt.title(f'Boxplot of {column}')
        plt.ylabel(column)
        plt.show()
```

```python
# Using plot to check for multicollinearity

plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.show()
```
